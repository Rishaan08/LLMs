```python
import os
from dotenv import load_dotenv

load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-120b", groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage
model.invoke([HumanMessage(content='Hi, My name is Rishaan and I am a learner.')])
```

```python
from langchain_core.messages import AIMessage
model.invoke(
    [
        HumanMessage(content='Hi, My name is Rishaan and I am a learner.'),
        AIMessage(content='Hello Rishaan! 👋 It’s great to meet you. How can I help you on your learning journey today?'),
        HumanMessage(content='Hey what is my name and what I do?')
    ]
)
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables import RunnableWithMessageHistory

store = {}

def get_session_history(session_id:str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
config = {"configurable": {"session_id": "chat1"}}
```

```python
response = with_message_history.invoke(
    [HumanMessage(content='Hi, My name is Rishaan and I am a learner.')], config=config
)
```

```python
response.content
```

```python
with_message_history.invoke(
    [HumanMessage(content="What's my name")], config=config
)
```

```python
# Change the config
config1 = {"configurable": {"session_id": "chat2"}}

with_message_history.invoke(
    [HumanMessage(content="What's my name")], config=config1
)
```

```python
response = with_message_history.invoke(
    [HumanMessage(content="My name is Kshitij.")], config=config1
)

response.content
```

```python
response = with_message_history.invoke(
    [HumanMessage(content="What's my name?")], config=config1
)

response.content
```

## Prompt Templates

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages(
    [
        ("system",
         "You are a helfpul assistant. Answer all the question to the best of your ability."),
        MessagesPlaceholder(variable_name='messages')
    ]
)

chain = prompt | model
```

```python
chain.invoke({"messages": [HumanMessage(content="Hi, My name is Rishaan.")]})
```

```python
with_message_history = RunnableWithMessageHistory(chain, get_session_history)
```

```python
config2 = {"configurable": {"session_id": "chat3"}}

response = with_message_history.invoke(
    [HumanMessage(content="Hi, My name is Rishaan.")],
    config=config2
)

response.content
```

```python
response = with_message_history.invoke(
    [HumanMessage(content="What's my name?")],
    config=config2
)

response.content
```

```python
# Add more complexity
prompt = ChatPromptTemplate.from_messages(
    [
        ("system",
         "You are a helfpul assistant. Answer all the question to the best of your ability in {language}."),
        MessagesPlaceholder(variable_name='messages')
    ]
)

chain = prompt | model
```

```python
response = chain.invoke({"messages": [HumanMessage(content="Hi, My name is Rishaan.")], "language":"Hindi"})
response.content
```

```python
with_message_history = RunnableWithMessageHistory(
    chain, get_session_history, input_messages_key="messages"
)
```

```python
config3 = {"configurable": {"session_id": "chat4"}}

response = with_message_history.invoke({"messages": [HumanMessage(content="Hi, I am Rishaan.")], "language":"Nepali"}, config=config3)

response.content
```

```python
response = with_message_history.invoke({"messages": [HumanMessage(content="What's my name?")], "language":"Nepali"}, config=config3)

response.content
```

## Manage the Conversation History

```python
from langchain_core.messages import SystemMessage, trim_messages

trimmer = trim_messages(
    max_tokens=45,
    strategy="last",
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"
)

messages = [
    SystemMessage(content="You're a good assistant."),
    HumanMessage(content="Hi, I am Dhriti."),
    AIMessage(content="Hi!"),
    HumanMessage(content="I like brownie."),
    AIMessage(content="Nice!"),
    HumanMessage(content="What is 368 + 632?"),
    AIMessage(content="Sum of both number is 1000"),
    HumanMessage(content="Nice talking to you."),
    AIMessage(content="Thank you always at your service.")
]

trimmer.invoke(messages)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough

chain = (
    RunnablePassthrough.assign(messages=itemgetter("messages")|trimmer)
    | prompt
    | model
)

response = chain.invoke(
    {
        "messages":messages + [HumanMessage(content="What is 368 + 632?")],
        "language": "English"
    }
)

response.content
```

```python
response = chain.invoke(
    {
        "messages":messages + [HumanMessage(content="Which sweet I like?")],
        "language": "English"
    }
)

response.content
```

```python
# Lets wrap this in message history
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)

config = {"configurable": {"session_id": "chat5"}}
```

```python
response = with_message_history.invoke(
    {
        "messages":messages + [HumanMessage(content="What's my name?'")],
        "language": "English"
    },
    config = config
)

response.content
```
